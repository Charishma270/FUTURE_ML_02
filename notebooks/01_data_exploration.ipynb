# 01 - Data Exploration

Goal: inspect the support ticket dataset, check quality issues, and understand category, priority, and ticket length distributions.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.append(str(ROOT / 'src'))

import matplotlib.pyplot as plt
import seaborn as sns
from utils import load_raw_data, prepare_ticket_dataframe
```

```python
raw_df = load_raw_data()
df = prepare_ticket_dataframe(raw_df)
df.head()
```

```python
print('Shape:', df.shape)
display(df.info())
display(df.isnull().sum().sort_values(ascending=False).head(12))
print('Duplicate rows:', df.duplicated().sum())
```

```python
display(df['category'].value_counts())
display(df['priority'].value_counts())
display(df['ticket_length'].describe())
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(data=df, y='category', order=df['category'].value_counts().index, color='#4c78a8')
plt.title('Category Count Plot')
plt.xlabel('Count')
plt.ylabel('Category')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='priority', order=df['priority'].value_counts().index, color='#f28e2b')
plt.title('Priority Count Plot')
plt.xlabel('Priority')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.histplot(df['ticket_length'], bins=30, color='#59a14f')
plt.title('Ticket Length Histogram')
plt.xlabel('Words per ticket')
plt.ylabel('Ticket count')
plt.tight_layout()
plt.show()
```

## Insights

- Category and priority distributions reveal whether the model may need class imbalance handling.
- Ticket length shows how much text signal is available for TF-IDF.
- Null and duplicate checks help avoid noisy training data and inflated evaluation metrics.
